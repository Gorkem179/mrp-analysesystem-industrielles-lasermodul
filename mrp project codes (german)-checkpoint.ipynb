```python
%%writefile kritiklik.py
import math

def faktor_umrechnen(punkte: int) -> float:
    """
    Wandelt Risikopunkte in Kritikalitätsfaktor um.
    Punkte 1-3  → 0.7  (geringes Risiko)
    Punkte 4-7  → 1.0  (mittleres Risiko)
    Punkte 8-10 → 1.5  (hohes Risiko)
    """
    if punkte >= 8:
        return 1.5
    elif punkte >= 4:
        return 1.0
    else:
        return 0.7

def kritikalitaetsfaktor(kritikalitaet: int, lieferrisiko: int, haltbarkeitsrisiko: int) -> float:
    """
    Berechnet den Kritikalitätsfaktor durch Multiplikation der drei Risikofaktoren.
    Warum Multiplikation statt gewichtetem Mittel?
    In der Hightech-Fertigung bestimmt das schwächste Glied das gesamte System.
    Diese Methodik ist identisch mit der FMEA-RPZ-Berechnung:
    RPZ = Bedeutung × Auftretenswahrscheinlichkeit × Entdeckungswahrscheinlichkeit
    """
    k1 = faktor_umrechnen(kritikalitaet)
    k2 = faktor_umrechnen(lieferrisiko)
    k3 = faktor_umrechnen(haltbarkeitsrisiko)
    return round(k1 * k2 * k3, 2)

def mindestbestand_berechnen(avg_dlz: int, max_dlz: int, wochenbedarf: int,
                              faktor: float) -> int:
    """
    Berechnet den Mindestbestand.
    Formel: (Max. DLZ - Avg. DLZ) × Wochenbedarf × Kritikalitätsfaktor
    Warum CEILING? Bauteile sind ganzzahlig.
    Kosten eines Produktionsstopps >> Kosten eines zusätzlichen Bauteils.
    """
    rohwert = (max_dlz - avg_dlz) * wochenbedarf * faktor
    return math.ceil(rohwert)

def bestandsstatus(aktueller_bestand: int, mindestbestand: int) -> str:
    if aktueller_bestand < mindestbestand:
        return "⚠ Kritischer Bestand"
    return "✔ Ausreichend"
```

```python
%%writefile mrp.py
import pandas as pd
import math
from kritiklik import kritikalitaetsfaktor, mindestbestand_berechnen, bestandsstatus

def daten_einlesen(dateipfad: str) -> tuple:
    """Liest die drei Arbeitsblätter aus der Excel-Datei."""
    stueckliste  = pd.read_excel(dateipfad, sheet_name="Stückliste",      header=1)
    lagerbestand = pd.read_excel(dateipfad, sheet_name="Lagerbestand",    header=1)
    produktplan  = pd.read_excel(dateipfad, sheet_name="Produktionsplan", header=1)

    stueckliste  = stueckliste.dropna(subset=["Ebene"])
    lagerbestand = lagerbestand.dropna(subset=["Teile-\nnummer"])
    produktplan  = produktplan.dropna(subset=["Woche"])

    return stueckliste, lagerbestand, produktplan

def lageranalyse(lager_df: pd.DataFrame) -> pd.DataFrame:
    """Berechnet Kritikalitätsfaktor, Mindestbestand und Status für jedes Bauteil."""
    lager_df = lager_df.drop(columns=["Unnamed: 12"], errors="ignore")
    ergebnisse = []

    for _, row in lager_df.iterrows():
        teilenr    = row["Teile-\nnummer"]
        bezeichnung= row["Bauteilbezeichnung"] if "Bauteilbezeichnung" in lager_df.columns else row.iloc[1]
        bestand    = int(row["Aktueller\nBestand"])
        avg_dlz    = int(row["Ø Durch-\nlaufzeit (Wo.)"])
        max_dlz    = int(row["Max. Durch-\nlaufzeit (Wo.)"])
        wochbed    = int(row["Wochen-\nbedarf"])
        krit       = int(row["Kritikalitäts-\nrisiko (1-10)"])
        liefer     = int(row["Liefer-\nrisiko (1-10)"])
        haltbar    = int(row["Haltbarkeits-\nrisiko (1-10)"])

        faktor     = kritikalitaetsfaktor(krit, liefer, haltbar)
        mindest    = mindestbestand_berechnen(avg_dlz, max_dlz, wochbed, faktor)
        status     = bestandsstatus(bestand, mindest)

        ergebnisse.append({
            "Teilenummer"          : teilenr,
            "Bauteilbezeichnung"   : bezeichnung,
            "Aktueller Bestand"    : bestand,
            "Ø DLZ (Wo.)"          : avg_dlz,
            "Max. DLZ (Wo.)"       : max_dlz,
            "Wochenbedarf"         : wochbed,
            "Kritikalitätsrisiko"  : krit,
            "Lieferrisiko"         : liefer,
            "Haltbarkeitsrisiko"   : haltbar,
            "Kritikalitätsfaktor"  : faktor,
            "Mindestbestand"       : mindest,
            "Status"               : status
        })
    return pd.DataFrame(ergebnisse)

def mrp_berechnen(stueck_df: pd.DataFrame, lager_df: pd.DataFrame,
                  produkt_df: pd.DataFrame) -> pd.DataFrame:
    """
    MRP-Hauptberechnung.
    Schritte:
    1. Nur Ebene-2-Bauteile (Rohmaterialien) verwenden
    2. Bruttobedarf = Bedarfsmenge × Menge pro Einheit
    3. Nettobedarf = Bruttobedarf - verfügbarer Bestand (unter Berücksichtigung Mindestbestand)
    4. Bestellwoche = Bedarfswoche - Durchlaufzeit
    """
    rohteile = stueck_df[stueck_df["Ebene"] == 2].copy()

    bestand_dict  = dict(zip(lager_df["Teile-\nnummer"], lager_df["Aktueller\nBestand"].astype(int)))
    dlz_dict      = dict(zip(lager_df["Teile-\nnummer"], lager_df["Ø Durch-\nlaufzeit (Wo.)"].astype(int)))
    mindest_dict  = {}

    for _, row in lager_df.iterrows():
        nr      = row["Teile-\nnummer"]
        faktor  = kritikalitaetsfaktor(
            int(row["Kritikalitäts-\nrisiko (1-10)"]),
            int(row["Liefer-\nrisiko (1-10)"]),
            int(row["Haltbarkeits-\nrisiko (1-10)"])
        )
        mindest_dict[nr] = mindestbestand_berechnen(
            int(row["Ø Durch-\nlaufzeit (Wo.)"]),
            int(row["Max. Durch-\nlaufzeit (Wo.)"]),
            int(row["Wochen-\nbedarf"]),
            faktor
        )

    bestand_aktuell = bestand_dict.copy()
    ergebnisse = []

    for _, woche_row in produkt_df.iterrows():
        woche   = int(woche_row["Woche"])
        bedarf  = int(woche_row["Bedarfsmenge (Stück)"])

        for _, teil_row in rohteile.iterrows():
            nr      = teil_row["Teilenummer"]
            menge   = int(teil_row["Menge pro Einheit"])

            if nr not in bestand_aktuell:
                continue

            bruttobedarf      = bedarf * menge
            verfuegbar        = max(0, bestand_aktuell[nr] - mindest_dict.get(nr, 0))
            nettobedarf       = max(0, bruttobedarf - verfuegbar)
            bestand_aktuell[nr] = max(0, bestand_aktuell[nr] - bruttobedarf)
            bestellwoche      = woche - dlz_dict.get(nr, 1)

            ergebnisse.append({
                "Woche"               : woche,
                "Teilenummer"         : nr,
                "Bauteilbezeichnung"  : str(teil_row["Bauteilbezeichnung"]).strip(),
                "Bruttobedarf"        : bruttobedarf,
                "Nettobedarf"         : nettobedarf,
                "Bestellwoche"        : bestellwoche,
                "Bestellung erforderlich": "Ja" if nettobedarf > 0 else "Nein"
            })

    return pd.DataFrame(ergebnisse)
```

```python
%%writefile kapasite.py
import pandas as pd

def kapazitaet_berechnen(produkt_df: pd.DataFrame,
                          taegl_arbeitszeit: float = 8.0,
                          arbeitstage_pw: int = 5,
                          wartungskapazitaet: float = 16.0,
                          zykluszeit: float = 3.5,
                          effizienz: float = 0.85) -> pd.DataFrame:
    """
    Führt die wöchentliche Kapazitätsanalyse durch.

    Taktzeit: Verfügbare Zeit pro Einheit zur Erfüllung der Kundennachfrage
              Taktzeit = Effektive Kapazität / Bedarfsmenge
    Zykluszeit: Tatsächliche Produktionszeit pro Einheit

    Regel:
    Zykluszeit < Taktzeit → Nachfrage wird erfüllt ✔
    Zykluszeit > Taktzeit → Engpass ⚠

    Ideale Kapazitätsauslastung: 70-85%
    < 70%  → Ungenutzte Kapazität (Kostenverschwendung)
    85-100% → Puffer schwindet (Achtung)
    > 100% → Engpass (Sofortmaßnahmen)
    """
    effektive_kap   = taegl_arbeitszeit * arbeitstage_pw * effizienz
    effektive_wart  = wartungskapazitaet * effizienz
    ergebnisse = []

    for _, row in produkt_df.iterrows():
        woche      = int(row["Woche"])
        bedarf     = int(row["Bedarfsmenge (Stück)"])
        bemerkung  = str(row["Bemerkung"])
        is_wartung = "Wartung" in bemerkung

        effektiv   = effektive_wart if is_wartung else effektive_kap
        taktzeit   = round(effektiv / bedarf, 2) if bedarf > 0 else 0
        erf_kap    = round(bedarf * zykluszeit, 1)
        differenz  = round(effektiv - erf_kap, 1)
        auslastung = round((erf_kap / effektiv) * 100, 1) if effektiv > 0 else 0

        if zykluszeit > taktzeit:
            takt_status = "⚠ Engpass — Zyklus > Takt"
        else:
            takt_status = "✔ Ausreichend — Zyklus < Takt"

        if is_wartung:          status = "🔵 Wartungswoche"
        elif auslastung > 100:  status = "🔴 Engpass"
        elif auslastung > 85:   status = "🟡 Achtung"
        elif auslastung >= 70:  status = "🟢 Idealbereich"
        else:                   status = "🔵 Ungenutzte Kapazität"

        ergebnisse.append({
            "Woche"                        : woche,
            "Bemerkung"                    : bemerkung,
            "Bedarfsmenge (Stück)"         : bedarf,
            "Effektive Kapazität (Std.)"   : effektiv,
            "Taktzeit (Std./Stück)"        : taktzeit,
            "Zykluszeit (Std./Stück)"      : zykluszeit,
            "Erf. Kapazität (Std.)"        : erf_kap,
            "Kapazitätsdifferenz (Std.)"   : differenz,
            "Kapazitätsauslastung (%)"     : f"%{auslastung}",
            "Taktzeit/Zykluszeit Status"   : takt_status,
            "Kapazitätsstatus"             : status
        })

    return pd.DataFrame(ergebnisse)

def kapazitaet_zusammenfassung(df: pd.DataFrame) -> dict:
    """Gibt zusammenfassende Statistiken der Kapazitätsanalyse zurück."""
    auslastung = df["Kapazitätsauslastung (%)"].str.replace("%", "").astype(float)
    return {
        "Durchschnittliche Auslastung" : f"%{auslastung.mean():.1f}",
        "Maximale Auslastung"          : f"%{auslastung.max():.1f}",
        "Minimale Auslastung"          : f"%{auslastung.min():.1f}",
        "Engpass-Wochen"               : int((auslastung > 100).sum()),
        "Unterauslastungs-Wochen"      : int((auslastung < 70).sum()),
        "Wochen im Idealbereich"       : int(((auslastung >= 70) & (auslastung <= 85)).sum()),
    }
```

```python
%%writefile main.py
import pandas as pd
from kritiklik import kritikalitaetsfaktor, mindestbestand_berechnen, bestandsstatus
from mrp import daten_einlesen, lageranalyse, mrp_berechnen
from kapasite import kapazitaet_berechnen, kapazitaet_zusammenfassung

DATEI = "mrp_project.xlsx"

print("=" * 60)
print("  Jenoptik – Lasermodul | MRP-Analysesystem")
print("=" * 60)

stueck_df, lager_df, produkt_df = daten_einlesen(DATEI)
print(f"Daten eingelesen:")
print(f"  Stückliste      : {len(stueck_df)} Einträge")
print(f"  Lagerbestand    : {len(lager_df)} Bauteile")
print(f"  Produktionsplan : {len(produkt_df)} Wochen")

print("\n-- Kritikalitäts- & Lageranalyse --")
lager_ergebnis   = lageranalyse(lager_df)
kritisch         = lager_ergebnis[lager_ergebnis["Status"] == "⚠ Kritischer Bestand"]
ausreichend      = lager_ergebnis[lager_ergebnis["Status"] == "✔ Ausreichend"]
print(f"Gesamt: {len(lager_ergebnis)} | Ausreichend: {len(ausreichend)} | Kritisch: {len(kritisch)}")

if len(kritisch) > 0:
    print("\n⚠ KRITISCHER BESTAND — SOFORTMASSNAHME ERFORDERLICH:")
    for _, row in kritisch.iterrows():
        print(f"  [{row['Teilenummer']}] {row['Bauteilbezeichnung']}")
        print(f"  Aktuell: {row['Aktueller Bestand']} | Mindestbestand: {row['Mindestbestand']}")

print("\n-- MRP-Berechnung --")
mrp_ergebnis     = mrp_berechnen(stueck_df, lager_df, produkt_df)
bestellung_erf   = mrp_ergebnis[mrp_ergebnis["Bestellung erforderlich"] == "Ja"]
print(f"Gesamt: {len(mrp_ergebnis)} | Bestellungen erforderlich: {len(bestellung_erf)}")

print("\n-- Kapazitätsplanung --")
kap_ergebnis = kapazitaet_berechnen(
    produkt_df,
    taegl_arbeitszeit  = 8.0,
    arbeitstage_pw     = 5,
    wartungskapazitaet = 16.0,
    zykluszeit         = 3.5,
    effizienz          = 0.85
)
zusammenfassung = kapazitaet_zusammenfassung(kap_ergebnis)
print("Kapazitätszusammenfassung:")
for k, v in zusammenfassung.items():
    print(f"  {k:<35}: {v}")

print("\n-- Excel-Ausgabe --")
ausgabe = "mrp_ergebnis.xlsx"
with pd.ExcelWriter(ausgabe, engine="openpyxl") as writer:
    lager_ergebnis.to_excel(writer, sheet_name="Lageranalyse",      index=False)
    mrp_ergebnis.to_excel(writer,   sheet_name="MRP_Ergebnis",      index=False)
    bestellung_erf.to_excel(writer, sheet_name="Bestellvorschlag",  index=False)
    kap_ergebnis.to_excel(writer,   sheet_name="Kapazitätsanalyse", index=False)

print(f"Gespeichert: {ausgabe}")
print(f"  Lageranalyse      : {len(lager_ergebnis)} Einträge")
print(f"  MRP_Ergebnis      : {len(mrp_ergebnis)} Einträge")
print(f"  Bestellvorschlag  : {len(bestellung_erf)} Einträge")
print(f"  Kapazitätsanalyse : {len(kap_ergebnis)} Einträge")
print("=" * 60)
print("  Analyse abgeschlossen.")
print("=" * 60)
```
